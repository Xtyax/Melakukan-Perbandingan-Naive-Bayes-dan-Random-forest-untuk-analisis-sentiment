```python
#Import library yang dibutuhkan
import csv
import nltk
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from nltk.tokenize import word_tokenize
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
```

```python
# Baca file excel
#, error_bad_lines=False
data_komentar = pd.read_excel('Data_Baru.xlsx')
data_komentar
```

```python
# Proses Cleaning Data dan Case Folding (Mengubah huruf menjadi kecil)
import re
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemover, ArrayDictionary
import string
def remove_excessive_repetitions(text):
    # Mengganti lebih dari dua pengulangan karakter dengan dua karakter
    return re.sub(r'(.)\1{2,}', r'\1\1', text)
def filtering_text(text):
    # mengubah komentar menjadi huruf kecil
    text = text.lower()
    # menghilangkan url
    text = re.sub(r'https?:\/\/\S+','',text)
    # menghilangkan mention, link, hastag
    text = ' '.join(re.sub("([@@#][A-Za-z0-9]+)|(\w+:\/\/\S+)"," ", text).split())
    #menghilangkan karakter byte (b')
    text = re.sub(r'(b\'{1,2})',"", text)
    # menghilangkan yang bukan huruf
    text = re.sub('[^a-zA-Z]', ' ', text)
    # menghilangkan digit angka
    text = re.sub(r'\d+', '', text)
    #menghilangkan tanda baca
    text = text.translate(str.maketrans("","",string.punctuation))
    # menghilangkan whitespace berlebih
    text = re.sub(r'\s+', ' ', text).strip()
    return text
```

```python
# Mengganti nilai NaN dengan string kosong dan memastikan semua nilai adalah string
data_komentar['text'] = data_komentar['text'].fillna('').astype(str)
```

```python
# Mencetak hasil dari cleaning dan case folding data
data_komentar['Case Folding'] = data_komentar['text'].apply(filtering_text)
data_komentar
```

```python
# Melakukan Tokenizing terhadap data
def word_tokenize_wrapper(text):
    return word_tokenize(text)



data_komentar['Tokenizing'] = data_komentar['Case Folding'].apply(word_tokenize_wrapper)
data_komentar
```

```python
# Melakukan Word Handling (Normalisasi)
# Kamus Slangs yang digunakan merupakan kamus yang diambil dari situs github 
# https://github.com/adeariniputri/text-preprocesing

kamus_normalisasi = pd.read_csv("slang.csv")

kata_normalisasi_dict = {}

for index, row in kamus_normalisasi.iterrows():
    if row[0] not in kata_normalisasi_dict:
        kata_normalisasi_dict[row[0]] = row[1] 
        
def normalisasi_kata(document):
    return [kata_normalisasi_dict[term] if term in kata_normalisasi_dict else term for term in document]

data_komentar['Word Handling'] = data_komentar['Tokenizing'].apply(normalisasi_kata)
data_komentar
```

```python
# Mengimport stopwords
# nltk.download('stopwords')
from nltk.corpus import stopwords

# Membuat agar variabel list_stopwords berisi stopwords berbahasa indonesia
list_stopwords = stopwords.words('indonesian')

# Membuat kamus stopwords 
list_stopwords = set(list_stopwords)

# Melakukan stopwords pada kata
def stopwords_removal(tokenized_text):
    return [word for word in tokenized_text if word not in list_stopwords]

data_komentar['Filtering'] = data_komentar['Word Handling'].apply(stopwords_removal)
data_komentar = data_komentar.astype(str)
data_komentar
```

```python
# Proses Stemming
def stemming(tokenized_text):
    # stemming
    factory_stemmer = StemmerFactory()
    stemmer = factory_stemmer.create_stemmer()
    text = stemmer.stem(tokenized_text)
    return text

data_komentar['Stemming'] = data_komentar['Filtering'].apply(stemming)
data_komentar = data_komentar.astype(str)
data_komentar
```

```python
data_komentar['Data Clean'] = data_komentar['Stemming']
data_komentar
```

```python
# Menyimpan data ke dalam file CSV dengan nama kolom sebagai header
data_komentar['Data Clean'].to_csv('data_clean.csv', index=False)
```
